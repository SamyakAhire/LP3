```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.layers import Dense,Conv1D, Flatten
from tensorflow.keras.models import Sequential, Model
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix, accuracy_score, precision_score, recall_score, PrecisionRecallDisplay, RocCurveDisplay
from sklearn.neural_network import MLPClassifier
```

```python
df = pd.read_csv('Churn_Modelling.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
plt.xlabel('Exited')
plt.ylabel('Count')
df['Exited'].value_counts().plot.bar()
plt.show()
```

```python
df['Geography'].value_counts()
```

```python
df = pd.concat([df,pd.get_dummies(df['Geography'],prefix='Geo')],axis=1)
```

```python
df = pd.concat([df,pd.get_dummies(df['Gender'])],axis=1)
```

```python
df.info()
```

```python
df.drop(columns=['RowNumber','Geography','Gender','Surname','CustomerId'],inplace=True)
```

```python
df.head()
```

```python
y = df['Exited'].values
x = df.loc[:,df.columns != 'Exited'].values
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=20)
```

```python
std_x = StandardScaler()
x_train = std_x.fit_transform(x_train)
x_test = std_x.transform(x_test)
```

```python
x_train.shape
```

```python
model = Sequential()
model.add(Flatten(input_shape=(13,)))
model.add(Dense(100,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.compile(optimizer='adam',metrics=['accuracy'],loss='BinaryCrossentropy')
```

```python
model.fit(x_train,y_train,batch_size=64,validation_split=0.1,epochs=100)
```

```python
print("X_train shape:", x_train.shape)
print("X_test shape:", x_test.shape)
```

```python
pred = model.predict(x_test)
```

```python
y_pred = []
for val in pred:
  if val > 0.5:
    y_pred.append(1)
  else:
    y_pred.append(0)
```

```python
accuracy_score(y_test,y_pred)
```

```python
cm = confusion_matrix(y_test,y_pred)
display = ConfusionMatrixDisplay(cm)
display.plot()
```

```python
nn_classifier = MLPClassifier(hidden_layer_sizes=(100),activation='logistic',max_iter=300,)
nn_classifier.fit(x_train,y_train)
```

```python
y_pred2 = nn_classifier.predict(x_test)
```

```python
accuracy_score(y_pred=y_pred2,y_true=y_test)
```

```python
nn_classifier.score(x_test,y_test)
```
